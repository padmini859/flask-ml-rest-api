```python
!pip install flask -q

from flask import Flask, jsonify
from sklearn.tree import DecisionTreeClassifier
import threading
import time
import requests

# ML model
X = [[1], [2], [3], [4], [5], [6]]
y = [0, 0, 0, 1, 1, 1]

model = DecisionTreeClassifier()
model.fit(X, y)

app = Flask(__name__)

@app.route('/')
def home():
    return "ML Model API is Running!"

@app.route('/predict/<int:value>')
def predict(value):
    prediction = model.predict([[value]])[0]
    return jsonify({
        "input": value,
        "prediction": int(prediction)
    })

# Run Flask in background
thread = threading.Thread(
    target=lambda: app.run(port=5000, use_reloader=False)
)
thread.daemon = True
thread.start()

time.sleep(2)

# Test API
response = requests.get("http://127.0.0.1:5000/predict/5")
print(response.json())
```
